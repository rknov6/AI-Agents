```python
spark.sql("""
          create or replace function agentic.agentic_schema.get_policy_details(policy_name string)
          returns table (
              policy string,
              policy_details string,
              last_updated string
          )
          language sql
          return (
              select policy,
                    policy_details,
                    last_updated
              from agentic.agentic_schema.policies
              where policy = policy_name
          )
          """
    )
```

```python
%sql
select  * from agentic.agentic_schema.get_policy_details("Refund Policy")
```

```python
spark.sql("""
          create or replace function agentic.agentic_schema.get_customer_details(user_email string)
          returns table (
              returns_12_months int,
              issue_category string,
              todays_date date
          )
          language sql
          return (
              select count(*) as returns_12_months,
              issue_category,
              now() as todays_date
              from agentic.agentic_schema.cust_service_data
              where email = user_email
              group by issue_category
          )
          """)
```
