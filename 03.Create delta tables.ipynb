```python
from pyspark.sql.functions import concat,col,lit
```

```python
files_path = '/Volumes/agentic/agentic_schema/files/01_Data_Files/'
target_path = 'agentic.agentic_schema.'
```

```python
(
    spark.read.csv(f"{files_path}products.csv",header = True)
              .writeTo(f'{target_path}products')
              .createOrReplace()
)
```

```python
(
    spark.read.csv(f"{files_path}policies.csv",header = True)
              .writeTo(f'{target_path}policies')
              .createOrReplace()
)
```

```python
(
    spark.read.csv(f"{files_path}cust_service_data.csv",header = True)
              .writeTo(f'{target_path}cust_service_data')
              .createOrReplace()
)
```

```python
products_df = spark.table(f'{target_path}products')
products_docs_df = spark.table(f'{target_path}product_docs')
```

```python

(
    products_df.alias("p")
               .join(products_docs_df.alias("pd"),"product_name","inner")
               .withColumn("indexed_column"
                           ,concat(
                               lit("<product_category>"),
                               col("product_category"),
                               lit("</product_category>\n"),
                               lit("<product_sub_category>"),
                               col("product_sub_category"),
                               lit("</product_sub_category>\n"),
                               lit("<product_name>"),
                               col("product_name"),
                               lit("</product_name>\n"),
                               lit("<product_doc>"),
                               col("product_doc"),
                               lit("</product_doc>\n")
                               ))
               .writeTo(f"{target_path}product_docs_combined")
               .createOrReplace()
)
```
